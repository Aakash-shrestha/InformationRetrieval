# Indexing Techniques

```python
documents = [
    "the cat sat on the mat",
    "the dog sat on the log",
    "the cat chased the dog",
    "dogs and cats can be friends"
]
```

```python
import re

def preprocess(text):
    # lowercase, remove punctuation, tokenize
    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text)
    return text.split()

tokenized_docs = [preprocess(doc) for doc in documents]
```

### Inverted Index

```python
from collections import defaultdict

inverted_index = defaultdict(list)

for doc_id, tokens in enumerate(tokenized_docs):
    for token in set(tokens):  # use set to avoid duplicates
        inverted_index[token].append(doc_id)

# Convert to regular dict for readability
inverted_index = dict(inverted_index)

print("Inverted Index:")
for term, doc_ids in inverted_index.items():
    print(f"{term}: {doc_ids}")
```

### Positional Inverted Index

```python
positional_index = defaultdict(lambda: defaultdict(list))

for doc_id, tokens in enumerate(tokenized_docs):
    for pos, token in enumerate(tokens):
        positional_index[token][doc_id].append(pos)

# Convert nested defaultdict to dict
positional_index = {term: dict(doc_pos) for term, doc_pos in positional_index.items()}

print("\nPositional Inverted Index:")
for term, doc_pos in positional_index.items():
    print(f"{term}: {doc_pos}")
```
